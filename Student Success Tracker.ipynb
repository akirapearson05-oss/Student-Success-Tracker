```python
import pandas as pd
import matplotlib.pyplot as plt

print("Student Success Tracker - Data Analysis")
```

```python
data = {
    "Student": ["S001", "S001", "S002", "S002", "S003", "S003"],
    "Course": ["Accounting", "Statistics", "Accounting", "Statistics", "Accounting", "Statistics"],
    "Average_Grade": [82, 75, 91, 86, 68, 72],
    "Study_Hours": [5, 4, 7, 6, 3, 4],
    "Missing_Assignments": [1, 2, 0, 1, 3, 2]
}

df = pd.DataFrame(data)

df
```

```python
course_averages = df.groupby("Course")["Average_Grade"].mean()

course_averages
```

```python
course_averages.plot(kind="bar")

plt.title("Average Grade by Course")
plt.xlabel("Course")
plt.ylabel("Average Grade (%)")
plt.ylim(0, 100)
plt.show()
```

```python
average_study_hours = df["Study_Hours"].mean()
average_grade = df["Average_Grade"].mean()
total_missing = df["Missing_Assignments"].sum()

print("Student Success Tracker Insights")
print("-------------------------------")
print(f"Average study hours: {average_study_hours:.1f} hours per week")
print(f"Overall average grade: {average_grade:.1f}%")
print(f"Total missing assignments: {total_missing}")
```

```python
plt.scatter(df["Study_Hours"], df["Average_Grade"])

plt.title("Study Hours vs. Average Grade")
plt.xlabel("Study Hours per Week")
plt.ylabel("Average Grade (%)")
plt.show()
```

```python
lowest_course = course_averages.idxmin()
lowest_grade = course_averages.min()

highest_course = course_averages.idxmax()
highest_grade = course_averages.max()

print("STUDENT SUCCESS INSIGHTS")
print("------------------------")
print(f"Course needing the most attention: {lowest_course}")
print(f"Lowest course average: {lowest_grade:.1f}%")
print(f"Highest-performing course: {highest_course}")
print(f"Highest course average: {highest_grade:.1f}%")
```

```python
missing_by_course = df.groupby("Course")["Missing_Assignments"].sum()

missing_by_course.plot(kind="bar")

plt.title("Missing Assignments by Course")
plt.xlabel("Course")
plt.ylabel("Number of Missing Assignments")
plt.show()
```

```python
print("PERSONALIZED RECOMMENDATION")
print("---------------------------")

if lowest_grade < 80:
    print(f"Consider spending extra study time on {lowest_course}.")
else:
    print("Keep up the good work in all courses!")

if total_missing > 0:
    print("Review missing assignments and make a plan to complete them.")
else:
    print("Great job keeping up with assignments!")
```
